```python
import pandas as pd

# Đọc file CSV
df = pd.read_csv("Hospital Wait  TIme Data.csv")

# Đếm số lượng missing từng cột và chỉ lấy cột > 0
missing_summary = df.isnull().sum()
missing_summary = missing_summary[missing_summary > 0].sort_values(ascending=False)

# Hiển thị kết quả
print("Các cột có dữ liệu thiếu:")
print(missing_summary)
```

```python
df["TestsOrdered"] = df["TestsOrdered"].fillna("None")
df["InsuranceType"] = df["InsuranceType"].fillna("None")

# Kiểm tra lại số lượng missing
print(df[["TestsOrdered", "InsuranceType"]].isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print(f"Số dòng dữ liệu bị trùng: {duplicate_count}")

# Hiển thị các dòng trùng (nếu có)
duplicates = df[df.duplicated()]

if duplicate_count > 0:
    print("\nCác dòng bị trùng:")
    print(duplicates)
else:
    print("\nKhông có dòng dữ liệu trùng.")
```

```python
# list tất cả cột datetime của bạn
datetime_cols = [
    'VisitDate',
    'AppointmentTime',
    'ActualArrivalTime',
    'CheckInTime',
    'TriageCompleteTime',
    'ProviderStartTime',
    'ProviderEndTime',
    'TestsCompleteTime',
    'DischargeTime',
    'ArrivalDateTime'
]

# convert tất cả về datetime chuẩn
for col in datetime_cols:
    df[col] = pd.to_datetime(df[col], format='mixed', dayfirst=True, errors='coerce')

# drop dòng lỗi datetime
df = df.dropna(subset=['ActualArrivalTime', 'DischargeTime'])

# OPTIONAL: đảm bảo logic thời gian đúng
df = df[df['DischargeTime'] >= df['ActualArrivalTime']]

# convert về format thống nhất (string đẹp để Tableau đọc)
for col in datetime_cols:
    df[col] = df[col].dt.strftime('%Y-%m-%d %H:%M:%S')

# lưu file sạch
df.to_csv("clean_hospital_data.csv", index=False)

print("DONE CLEAN DATA")
```

```python
import pandas as pd

# đọc file
df = pd.read_csv("clean_hospital_data.csv")

# convert datetime
df['ActualArrivalTime'] = pd.to_datetime(df['ActualArrivalTime'])
df['DischargeTime'] = pd.to_datetime(df['DischargeTime'])

# tạo cột ngày
df['VisitDate'] = df['ActualArrivalTime'].dt.date

hours = list(range(24))
results = []

# loop từng ngày
for date in df['VisitDate'].unique():
    df_day = df[df['VisitDate'] == date]

    for h in hours:
        hour_start = pd.Timestamp(date) + pd.Timedelta(hours=h)
        hour_end = hour_start + pd.Timedelta(hours=1)

        # 🔥 LOGIC MỚI: OVERLAP INTERVAL
        count = df_day[
            (df_day['ActualArrivalTime'] < hour_end) &
            (df_day['DischargeTime'] >= hour_start)
        ]['PatientID'].nunique()

        results.append({
            'VisitDate': date,
            'Hour': h,
            'Patients': count
        })

# dataframe kết quả
result_df = pd.DataFrame(results)

# trung bình theo giờ
avg_per_hour = result_df.groupby('Hour')['Patients'].mean().reset_index()

print(avg_per_hour)
```

```python
import pandas as pd

# 1. Chuẩn hóa ngày
df['Date'] = pd.to_datetime(df['VisitDate']).dt.date

# 2. Lấy thứ trong tuần
df['DayOfWeek'] = pd.to_datetime(df['VisitDate']).dt.day_name()

# 3. Đếm số bệnh nhân mỗi ngày
daily_counts = (
    df.groupby('Date')['PatientID']
    .nunique()   # tránh duplicate
    .reset_index(name='Daily Patients')
)

# 4. Gắn lại DayOfWeek
daily_counts['DayOfWeek'] = pd.to_datetime(daily_counts['Date']).dt.day_name()

# 5. Tính trung bình theo DayOfWeek
avg_by_dow = (
    daily_counts.groupby('DayOfWeek')['Daily Patients']
    .mean()
    .reset_index(name='Avg Patients')
)

# 6. Sort thứ
order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
avg_by_dow['DayOfWeek'] = pd.Categorical(avg_by_dow['DayOfWeek'], categories=order, ordered=True)
avg_by_dow = avg_by_dow.sort_values('DayOfWeek')

avg_by_dow
```

```python
import pandas as pd

# đọc file
df = pd.read_csv("clean_hospital_data.csv")

# convert datetime
df['ActualArrivalTime'] = pd.to_datetime(df['ActualArrivalTime'])
df['DischargeTime'] = pd.to_datetime(df['DischargeTime'])

# tạo cột ngày + giờ
df['VisitDate'] = df['ActualArrivalTime'].dt.date
df['ArrivalHour'] = df['ActualArrivalTime'].dt.hour

# ==============================
# 🔵 1. PATIENTS IN SYSTEM
# ==============================

hours = list(range(24))
results = []

for date in df['VisitDate'].unique():
    df_day = df[df['VisitDate'] == date]

    for h in hours:
        hour_start = pd.Timestamp(date) + pd.Timedelta(hours=h)
        hour_end = hour_start + pd.Timedelta(hours=1)

        # overlap interval
        count = df_day[
            (df_day['ActualArrivalTime'] < hour_end) &
            (df_day['DischargeTime'] >= hour_start)
        ]['PatientID'].nunique()

        results.append({
            'VisitDate': date,
            'Hour': h,
            'Patients': count
        })

patients_df = pd.DataFrame(results)

# avg patients theo hour
avg_patients = patients_df.groupby('Hour')['Patients'].mean().reset_index()
avg_patients.rename(columns={'Patients': 'Avg_Patients'}, inplace=True)

# ==============================
# 🔵 2. PROVIDERS
# ==============================

tmp = df.groupby(["VisitDate", "ArrivalHour"])["ProviderID"].nunique().reset_index()

avg_providers = tmp.groupby("ArrivalHour")["ProviderID"].mean().reset_index()
avg_providers.rename(columns={
    "ArrivalHour": "Hour",
    "ProviderID": "Avg_Providers"
}, inplace=True)

# ==============================
# 🔵 3. MERGE + RATIO
# ==============================

final = pd.merge(avg_patients, avg_providers, on="Hour", how="inner")

# ratio
final["Patients_per_Provider"] = final["Avg_Patients"] / final["Avg_Providers"]

# ==============================
# 🔵 4. EXPORT FILE
# ==============================

final.to_csv("patients_vs_providers_hour.csv", index=False)

print(final)
```
